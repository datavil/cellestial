```python
import polars as pl
from lets_plot import *

LetsPlot.setup_html()
```

```python
df = pl.read_ipc("benchmarks/results.feather")
df =df.sort("library",descending=True)
#df = df.filter(pl.col("dataset") != "atlas200k")
df.head()
```

```python
(
    ggplot(df)
    + geom_boxplot(aes(x="library",y="time(s)", fill="library"))
    + facet_wrap("case")
    # + scale_y_log10()
    + theme_bw()
    # + ggsize(1400,800)
)
```

```python
(
    ggplot(df)
    + geom_point(aes(x="n_items", y="time(s)", color="library"), size=3)
    + geom_smooth(aes(x="n_items", y="time(s)", color="library"))
)
```
