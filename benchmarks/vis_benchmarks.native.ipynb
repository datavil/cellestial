```python
import polars as pl
from lets_plot import *

LetsPlot.setup_html()
```

```python
df = pl.read_ipc("benchmarks/results.native.feather")
df =df.sort("library",descending=True)
#df = df.filter(pl.col("dataset") != "atlas200k")
df.head()
```

```python
(
    ggplot(df)
    + geom_boxplot(aes(x="library",y="time(s)", fill="library"))
    + facet_wrap("case")
    # + scale_y_log10()
    + theme_bw()
    # + ggsize(1400,800)
)
```

```python
(
    ggplot(df)
    + geom_point(aes(x="n_items", y="time(s)", color="library"), size=3)
    + geom_smooth(aes(x="n_items", y="time(s)", color="library"))
)
```

```python
case_order = ["umap", "spatial", "violin","dotplot", "matrixplot",  "heatmap"]
df = df.with_columns(pl.col("case").cast(pl.Enum(case_order)))

(
    ggplot(df)
    + geom_boxplot(aes(x="library", y="time(s)", fill="library"), color="#1f1f1f")
    + theme_bw()
    + theme(axis_title_x=element_blank())
    + facet_wrap("case", order=0)
    + labs(fill="Library", y="Time (s)")
    + scale_y_log10()
)
```

```python
df2 = pl.read_ipc("benchmarks/results.feather")
df2 =df2.sort("library",descending=True)
#df = df.filter(pl.col("dataset") != "atlas200k")
df2.head()
```
