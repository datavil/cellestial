```python
import scanpy as sc
from lets_plot import *

import cellestial as cl

LetsPlot.setup_html()

data = sc.read("data/pbmc3k_pped.h5ad")
```

```python
dim = cl.dimensional(
    data,
    dimensions="umap",
    key="leiden",
    size=1,
    axis_type=None,
    tooltips=["leiden", "cell_type_lvl1", "n_genes", "total_counts_hb", "MT-ND2"],
    legend_ondata=True,
    ondata_size=12,
    ondata_color="black",
    ondata_fontface="bold",
    ondata_family="sans",
    ondata_alpha=0.8,
)
dim + ggsize(800, 600)
```

```python
(dim + facet_wrap(facets=["sample"]) + ggsize(600, 900))
```

```python
dim2 = cl.dimensional(
    data,
    dimensions="umap",
    key="MT-ND2",
    size=1,
    axis_type=None,
    color_high=cl.TEAL,
    tooltips=["leiden", "cell_type_lvl1", "n_genes", "total_counts_hb", "MT-ND2"],
    legend_ondata=True,
    ondata_size=12,
    ondata_color="black",
    ondata_fontface="bold",
    ondata_family="sans",
    ondata_alpha=0.8,
)
dim2 += ggsize(800, 600)
dim2
```
