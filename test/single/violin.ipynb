```python
import scanpy as sc
from lets_plot import *

import cellestial as cl

LetsPlot.setup_html()

data = sc.read("data/pbmc3k_pped.h5ad")
```

```python
cl.violin(
    data,
    "n_genes_by_counts",
    geom_color="#3f3f3f",
    # show_points=True,
    alpha=0.1,
    show_tooltips=True,
)
```

```python
cl.violin(
    data,
    "n_genes_by_counts",
    geom_color="#3f3f3f",
    fill="sample",
    # show_points=True,
    alpha=0.1,
    show_tooltips=True,
)
```

```python
base = cl.violin(
    data,
    "n_genes_by_counts",
    fill="sample",
    color="sample",
    geom_color="#3f3f3f",
    # show_points=True,
    alpha=0.1,
    show_tooltips=True,
)
base
```

```python
ab = cl.retrieve(base)
ab
```

```python
(
    base
    + labs(title="Cell type distribution", x="Samples", y="Number of Genes by Counts")
    + ggsize(600, 400)
)
```

```python
(
    cl.violin(data, "n_genes_by_counts", fill="leiden", point_geom="jitter")
    + ggsize(800, 400)
    + theme(
        title=element_text(family="Arial", color="#3f3f3f"),
    )
)
```

```python
cl.violins(
    data,
    ["n_genes_by_counts", "pct_counts_in_top_100_genes", "log1p_total_counts_mt", "pct_counts_hb"],
    ncol=2,
    fill="sample",
    layers=[scale_y_log10()],
)
```

```python
(
    cl.violin(
        data,
        "MT-ND5",
        fill="leiden",
        add_keys="sample",
    )
    + facet_wrap("sample")
    + ggsize(600, 800)
)
```

```python
abc = (
    cl.violin(
        data,
        key=["pct_counts_in_top_200_genes", "n_genes_by_counts"],
        trim=False,
        fill="sample",
    )
    + scale_y_log2()
    + ggsize(600, 400)
)
abc
```
