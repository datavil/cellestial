```python
import cellestial as cl
import scanpy as sc
from lets_plot import *

LetsPlot.setup_html()
data = sc.read("data/pbmc3k_pped.h5ad")
```

```python
cl.umap(data, key=None, color="gray", size=0.6)
```

## UMAP

```python
umap = cl.umap(
    data, key="leiden", size=1, axis_type="arrow", alpha=1, legend_ondata=True, interactive=True
) + ggsize(700, 600)
umap
```

```python
umap + scale_color_viridis(option="B")
```
