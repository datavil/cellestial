```python
import cellestial as cl
import scanpy as sc
from lets_plot import *

LetsPlot.setup_html()

data = sc.read("data/pbmc3k_pped.h5ad")
```

## A dimensional plot

```python
data.var_names
```

```python

dim = cl.dimensional(
    data,
    dimensions="umap",
    key="leiden",
    size=1,
    axis_type="arrow",
    alpha=0.6,
    #variable_keys=["n_genes", "total_counts_hb","MT-ND2"],
    tooltips=["n_genes", "total_counts_hb","MT-ND2"],
    legend_ondata=True,
    ondata_size=12,
    ondata_color="black",
    ondata_fontface="bold",
    ondata_family="sans",
    ondata_alpha=0.8,
)
dim
```

notes: x y axis titles are not placed accordingly

## Interactive

```python
dimI = cl.dimensional(data, interactive=True, key="leiden", dimensions="umap")
dimI
```

## Customizing Layers

### Size

```python
dimI += ggsize(700, 550)
dimI
```

### Titles

```python
dim += ggtitle("Dimensional reduction")
dim
```

works well

### Palettes

```python
dim += scale_color_hue()
dim
```

### Axis titles

```python
dim += theme(axis_title=element_text(size=12))
dim
```

### Legend

```python
dim += guides(color=guide_legend(ncol=3)) + ggsize(600, 400)
dim
```

### Reverse Axis

```python
dim += scale_y_reverse()
dim
```

notes: this does not work properly (0.1.2)
