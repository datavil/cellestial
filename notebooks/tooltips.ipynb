```python
import scanpy as sc
from lets_plot import layer_tooltips
from cellestial.util import _resolve_tooltips, _tooltip_fields

data = sc.read_h5ad('data/pbmc3k_pped.h5ad')
gene_a = data.var_names[0]

spec = layer_tooltips([gene_a]).line('Cluster|@leiden').format('^color', '.2f')
var_keys = []
resolved = _resolve_tooltips(spec, data=data, variable_keys=var_keys, defaults=[])
print(f'fields={_tooltip_fields(spec)}')
print(f'var_keys={var_keys}')
print(f'resolved is spec: {resolved is spec}')

var_keys = []
resolved = _resolve_tooltips(None, data=data, variable_keys=var_keys, defaults=['Barcode', gene_a])
print(f'None case -> resolved={resolved}, var_keys={var_keys}')

var_keys = []
resolved = _resolve_tooltips('none', data=data, variable_keys=var_keys, defaults=['Barcode'])
print(f'\"none\" case -> resolved={resolved!r}, var_keys={var_keys}')
```
