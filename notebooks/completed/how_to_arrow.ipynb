```python
import cellestial as cl
import scanpy as sc
from lets_plot import *
import numpy as np

data = sc.read("data/pbmc3k_mini.h5ad")
```

```python
base = cl.umap(data, axis_type="axis")
base
```

```python
x_min = data.obsm["X_umap"][:, 0].min()
y_min = data.obsm["X_umap"][:, 1].min()
```

```python
x_min, y_min
```

```python
x_breaks = np.linspace(x_min * 1.1, 1, 10).astype(int).tolist()
y_breaks = np.linspace(y_min * 1.1, 1, 10).astype(int).tolist()
```

```python
x_breaks
```

```python
base += theme(axis_ticks=element_line(color="#3f3f3f", size=25), axis_text=element_blank())
```

```python
base += scale_x_continuous(breaks=x_breaks, expand=(0, 0)) + scale_y_continuous(
    breaks=y_breaks, expand=(0, 0)
)
base
```
