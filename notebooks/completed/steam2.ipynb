```python
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream(
    dim,
    n_grid: int = 30,
    density: float = 1.0,
    color: str = "black",
    linewidth: float = 0.5,
    arrow_size: float = 6,
    min_mass: float = 1.0,
    integration_steps: int = 60,
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.05,
    arrow_every: int = 10,
    min_seed_dist_frac: float = 0.03,
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : grid resolution for velocity interpolation (default 30)
    density             : fraction of candidate seeds to use (default 1.0)
    color               : line and arrowhead colour (default "black")
    linewidth           : path line width (default 0.5)
    arrow_size          : arrowhead size in px (default 6)
    min_mass            : min kernel-density threshold — higher removes more
                          empty-region seeds (default 1.0)
    integration_steps   : max ODE steps per streamline (default 60)
    step_size           : ODE step size; defaults to ~80% of grid cell size
    max_cell_dist_frac  : streamline is cut when it wanders further than this
                          fraction of the data range from any real cell (default 0.05)
    arrow_every         : place an arrowhead every N path points (default 10)
    min_seed_dist_frac  : minimum distance between seeds as fraction of data
                          range, prevents overcrowding (default 0.03)

    Usage
    -----
    dim + cl.stream(dim)
    dim + cl.stream(dim, n_grid=40, density=1.5, color="black")
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names from the plot object
    # ------------------------------------------------------------------ #

    frame = retrieve(dim)
    mapping = get_mapping(dim)

    x_col = mapping["x"]  # e.g. "X_UMAP1"
    y_col = mapping["y"]  # e.g. "X_UMAP2"
    prefix = "VELOCITY_"
    vx_col = f"{prefix}{x_col.replace('X_', '')}"  # e.g. "VELOCITY_UMAP1"
    vy_col = f"{prefix}{y_col.replace('X_', '')}"  # e.g. "VELOCITY_UMAP2"

    x = frame[x_col].to_numpy().astype(np.float64)
    y = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Build regular grid and interpolate velocities
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    pad_x = x_range * 0.02
    pad_y = y_range * 0.02
    x_min, x_max = x.min() - pad_x, x.max() + pad_x
    y_min, y_max = y.min() - pad_y, y.max() + pad_y

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])  # (n_grid^2, 2)

    # Gaussian kernel bandwidth ~ 1.5 grid cells
    bw_x = x_range / n_grid * 1.5
    bw_y = y_range / n_grid * 1.5

    # Kernel density at each grid point
    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    # Weighted velocity interpolation onto the grid
    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    # Normalise to uniform speed so slow regions don't stall
    speed = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    # scipy interpolators — note (row=y, col=x) axis convention
    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])  # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Seed selection: grid points inside the data manifold,
    #     filtered by KDE mass + minimum inter-seed distance pruning
    # ------------------------------------------------------------------ #
    mass_flat = mass.ravel()
    mass_threshold = np.percentile(mass_flat, max(0, 100 - min_mass * 10))
    candidate_mask = mass_flat > mass_threshold
    candidate_pts = grid_pts[candidate_mask]

    # Sub-sample by density
    n_want = max(1, int(len(candidate_pts) * min(density, 1.0)))
    rng = np.random.default_rng(0)
    idx = rng.choice(len(candidate_pts), size=min(n_want, len(candidate_pts)), replace=False)
    candidate_pts = candidate_pts[idx]

    # Greedy minimum-distance pruning: ensures even spread, no clumping
    min_dist = min(x_range, y_range) * min_seed_dist_frac
    selected = []
    if len(candidate_pts) > 0:
        selected.append(candidate_pts[0])
        for pt in candidate_pts[1:]:
            dists = np.linalg.norm(np.array(selected) - pt, axis=1)
            if dists.min() >= min_dist:
                selected.append(pt)
    seeds = np.array(selected)

    # KD-tree of real cells — used to cut paths that leave the manifold
    cell_tree = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward through the velocity field
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt = step_size if step_size is not None else cell_size * 0.8
    t_span = (0.0, dt * integration_steps)
    t_eval = np.linspace(*t_span, integration_steps + 1)

    path_rows = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field,
                t_span,
                [seed[0], seed[1]],
                t_eval=t_eval,
                method="RK45",
                rtol=1e-3,
                atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        # Cut path at first point that strays too far from any real cell
        pts_arr = np.column_stack([xs, ys])
        dists, _ = cell_tree.query(pts_arr)
        # searchsorted finds the first True (out-of-bounds) index
        valid_end = (
            int(np.argmax(dists > max_cell_dist)) if np.any(dists > max_cell_dist) else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)

        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        # Arrowheads: every `arrow_every` steps, using consecutive points
        # so each arrow is perfectly tangent to the path
        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        # Always add one near the tip
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append(
                {
                    "x": float(xs[i]),
                    "y": float(ys[i]),
                    "xend": float(xs[i + 1]),
                    "yend": float(ys[i + 1]),
                }
            )

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data = {c: path_df[c].to_list() for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=0.8,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
```

```python
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream2(
    dim,
    n_grid: int = 30,
    density: float = 1.0,
    color: str = "black",
    linewidth: float = 0.5,
    arrow_size: float = 3,  # smaller: was 6, arrows were dominating
    min_mass: float = 2.0,  # stricter: was 1.0, seeds empty regions
    integration_steps: int = 60,
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.04,  # slightly tighter than 0.05
    arrow_every: int = 12,  # less frequent: was 10, reduces clutter
    min_seed_dist_frac: float = 0.05,  # wider spacing: was 0.03, reduces crowding
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : grid resolution for velocity interpolation (default 30)
    density             : fraction of candidate seeds to use (default 1.0)
    color               : line and arrowhead colour (default "black")
    linewidth           : path line width (default 0.5)
    arrow_size          : arrowhead size in px (default 3)
    min_mass            : min kernel-density threshold — higher removes more
                          empty-region seeds (default 2.0)
    integration_steps   : max ODE steps per streamline (default 60)
    step_size           : ODE step size; defaults to ~80% of grid cell size
    max_cell_dist_frac  : streamline is cut when it wanders further than this
                          fraction of the data range from any real cell (default 0.04)
    arrow_every         : place an arrowhead every N path points (default 12)
    min_seed_dist_frac  : minimum distance between seeds as fraction of data
                          range, prevents overcrowding (default 0.05)

    Usage
    -----
    dim + cl.stream(dim)
    dim + cl.stream(dim, n_grid=40, density=1.5, color="black")

    Tuning tips
    -----------
    Too crowded on one side  → increase min_seed_dist_frac (e.g. 0.07)
    Lines leaving data cloud → lower max_cell_dist_frac (e.g. 0.03)
    Too few lines overall    → lower min_mass (e.g. 1.0) or increase density
    Arrowheads too large     → lower arrow_size (e.g. 2)
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names from the plot object
    # ------------------------------------------------------------------ #
    frame = retrieve(dim)
    mapping = get_mapping(dim)

    x_col = mapping["x"]  # e.g. "X_UMAP1"
    y_col = mapping["y"]  # e.g. "X_UMAP2"
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"  # e.g. "VELOCITY_UMAP1"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"  # e.g. "VELOCITY_UMAP2"

    x = frame[x_col].to_numpy().astype(np.float64)
    y = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Build regular grid and interpolate velocities
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    pad_x = x_range * 0.02
    pad_y = y_range * 0.02
    x_min, x_max = x.min() - pad_x, x.max() + pad_x
    y_min, y_max = y.min() - pad_y, y.max() + pad_y

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])  # (n_grid^2, 2)

    # Gaussian kernel bandwidth ~ 1.5 grid cells
    bw_x = x_range / n_grid * 1.5
    bw_y = y_range / n_grid * 1.5

    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    # Full normalisation to uniform speed — every line moves, none stalls
    speed = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    # scipy interpolators — (row=y, col=x) axis convention
    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])  # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Seed selection: dense regions only, evenly spread
    # ------------------------------------------------------------------ #
    mass_flat = mass.ravel()
    mass_threshold = np.percentile(mass_flat, max(0, 100 - min_mass * 10))
    candidate_mask = mass_flat > mass_threshold
    candidate_pts = grid_pts[candidate_mask]

    n_want = max(1, int(len(candidate_pts) * min(density, 1.0)))
    rng = np.random.default_rng(0)
    idx = rng.choice(len(candidate_pts), size=min(n_want, len(candidate_pts)), replace=False)
    candidate_pts = candidate_pts[idx]

    # Greedy minimum-distance pruning — even spread, no regional clumping
    min_dist = min(x_range, y_range) * min_seed_dist_frac
    selected = []
    if len(candidate_pts) > 0:
        selected.append(candidate_pts[0])
        for pt in candidate_pts[1:]:
            if np.linalg.norm(np.array(selected) - pt, axis=1).min() >= min_dist:
                selected.append(pt)
    seeds = np.array(selected)

    # KD-tree for manifold clipping
    cell_tree = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt = step_size if step_size is not None else cell_size * 0.8
    t_span = (0.0, dt * integration_steps)
    t_eval = np.linspace(*t_span, integration_steps + 1)

    path_rows = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field,
                t_span,
                [seed[0], seed[1]],
                t_eval=t_eval,
                method="RK45",
                rtol=1e-3,
                atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        # Clip at first point that leaves the data manifold
        dists, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists > max_cell_dist)) if np.any(dists > max_cell_dist) else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        # Arrowheads perfectly tangent to the path at every arrow_every-th step
        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append(
                {
                    "x": float(xs[i]),
                    "y": float(ys[i]),
                    "xend": float(xs[i + 1]),
                    "yend": float(ys[i + 1]),
                }
            )

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data = {c: path_df[c].to_list() for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=0.8,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
```

```python
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream3(
    dim,
    n_grid: int = 30,
    density: float = 1.0,
    color: str = "black",
    linewidth: float = 0.5,
    arrow_size: float = 3,
    min_mass: float = 2.0,
    integration_steps: int = 60,
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.04,
    arrow_every: int = 12,
    min_seed_dist_frac: float = 0.05,
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    Seed density is log-normalised across the UMAP so sparse regions
    (e.g. a small cluster) get proportionally as many streamlines as dense ones.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : grid resolution for velocity interpolation (default 30)
    density             : total number of seeds to attempt (default 1.0 = auto)
    color               : line and arrowhead colour (default "black")
    linewidth           : path line width (default 0.5)
    arrow_size          : arrowhead size in px (default 3)
    min_mass            : min kernel-density threshold — higher removes more
                          empty-region seeds (default 2.0)
    integration_steps   : max ODE steps per streamline (default 60)
    step_size           : ODE step size; defaults to ~80% of grid cell size
    max_cell_dist_frac  : streamline is cut when it wanders further than this
                          fraction of the data range from any real cell (default 0.04)
    arrow_every         : place an arrowhead every N path points (default 12)
    min_seed_dist_frac  : minimum distance between seeds as fraction of data
                          range, prevents overcrowding (default 0.05)

    Tuning tips
    -----------
    Too crowded on one side  → increase min_seed_dist_frac (e.g. 0.07)
    Lines leaving data cloud → lower max_cell_dist_frac (e.g. 0.03)
    Too few lines overall    → lower min_mass (e.g. 1.0) or increase density
    Arrowheads too large     → lower arrow_size (e.g. 2)
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names from the plot object
    # ------------------------------------------------------------------ #
    frame = retrieve(dim)
    mapping = get_mapping(dim)

    x_col = mapping["x"]
    y_col = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    x = frame[x_col].to_numpy().astype(np.float64)
    y = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Regular grid + velocity interpolation
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    x_min, x_max = x.min() - x_range * 0.02, x.max() + x_range * 0.02
    y_min, y_max = y.min() - y_range * 0.02, y.max() + y_range * 0.02

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])

    bw_x = x_range / n_grid * 1.5
    bw_y = y_range / n_grid * 1.5

    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    speed = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])  # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Log-normalised seed sampling
    #
    #     Instead of sampling proportional to mass (which over-seeds dense
    #     regions), we draw seeds with probability proportional to
    #     log(1 + mass).  This compresses the dynamic range so a cluster
    #     with 10× more cells still gets a comparable number of streamlines.
    # ------------------------------------------------------------------ #
    mass_flat = mass.ravel()

    # Hard floor: discard grid points below the KDE threshold entirely
    mass_threshold = np.percentile(mass_flat, max(0, 100 - min_mass * 10))
    valid_mask = mass_flat > mass_threshold
    valid_pts = grid_pts[valid_mask]
    valid_mass = mass_flat[valid_mask]

    if len(valid_pts) == 0:
        raise RuntimeError("cl.stream(): no valid seed points found. Lower min_mass.")

    # Log-normalised sampling weights
    log_weights = np.log1p(valid_mass)
    log_weights /= log_weights.sum()

    n_seeds = max(1, int(len(valid_pts) * min(density, 1.0)))
    rng = np.random.default_rng(0)
    chosen = rng.choice(
        len(valid_pts), size=min(n_seeds, len(valid_pts)), replace=False, p=log_weights
    )
    candidate_pts = valid_pts[chosen]

    # Greedy minimum-distance pruning so seeds are spatially even
    min_dist = min(x_range, y_range) * min_seed_dist_frac
    selected = [candidate_pts[0]]
    for pt in candidate_pts[1:]:
        if np.linalg.norm(np.array(selected) - pt, axis=1).min() >= min_dist:
            selected.append(pt)
    seeds = np.array(selected)

    # KD-tree for manifold clipping
    cell_tree = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt = step_size if step_size is not None else cell_size * 0.8
    t_span = (0.0, dt * integration_steps)
    t_eval = np.linspace(*t_span, integration_steps + 1)

    path_rows = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field,
                t_span,
                [seed[0], seed[1]],
                t_eval=t_eval,
                method="RK45",
                rtol=1e-3,
                atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists > max_cell_dist)) if np.any(dists > max_cell_dist) else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append(
                {
                    "x": float(xs[i]),
                    "y": float(ys[i]),
                    "xend": float(xs[i + 1]),
                    "yend": float(ys[i + 1]),
                }
            )

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data = {c: path_df[c].to_list() for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=0.8,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
```

```python
from polars.selectors import alpha
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream4(
    dim,
    n_grid: int = 30,
    density: float = 1.0,
    color: str = "#1f1f1f",
    linewidth: float = 1,
    line_alpha: float = 0.6,
    arrow_size: float = 3,
    min_mass: float = 4.0,           # higher: filters truly empty space only
    integration_steps: int = 60,
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.05,
    arrow_every: int = 12,
    seeds_per_tile: int = 1,         # how many seeds to place per spatial tile
    n_tiles: int = 10,               # tile grid resolution (n_tiles x n_tiles)
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    Seeding is spatially uniform: the UMAP space is divided into equal-area
    tiles and each occupied tile gets exactly `seeds_per_tile` seeds regardless
    of local cell density.  This prevents dense regions from dominating.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : grid resolution for velocity interpolation (default 30)
    density             : scales the total seed count (default 1.0)
    color               : line and arrowhead colour (default "black")
    linewidth           : path line width (default 0.5)
    arrow_size          : arrowhead size in px (default 3)
    min_mass            : KDE threshold — grid points below this are not seeded
                          (default 4.0; lower to include sparser regions)
    integration_steps   : max ODE steps per streamline (default 60)
    step_size           : ODE step size; defaults to ~80 % of grid cell size
    max_cell_dist_frac  : streamline is cut when it wanders further than this
                          fraction of the data range from any real cell (0.04)
    arrow_every         : place an arrowhead every N path points (default 12)
    seeds_per_tile      : seeds placed per spatial tile (default 1)
    n_tiles             : spatial tile grid resolution (default 10 → 10×10 tiles)

    Tuning tips
    -----------
    Too few lines in sparse regions → lower min_mass (e.g. 2.0)
    Too many lines in dense regions → lower seeds_per_tile or raise n_tiles
    Lines leaving data cloud        → lower max_cell_dist_frac (e.g. 0.03)
    Arrowheads too large/frequent   → lower arrow_size / raise arrow_every
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names from the plot object
    # ------------------------------------------------------------------ #
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    x  = frame[x_col].to_numpy().astype(np.float64)
    y  = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Regular grid + velocity interpolation
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    x_min, x_max = x.min() - x_range * 0.02, x.max() + x_range * 0.02
    y_min, y_max = y.min() - y_range * 0.02, y.max() + y_range * 0.02

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY  = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])

    bw_x = x_range / n_grid * 1.5
    bw_y = y_range / n_grid * 1.5

    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w  = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    speed   = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Spatially uniform seeding via equal-area tiles
    #
    #     Divide the bounding box into n_tiles × n_tiles equal tiles.
    #     For each tile, collect all grid points that:
    #       (a) fall inside the tile, and
    #       (b) have KDE mass > min_mass threshold
    #     Then pick exactly `seeds_per_tile` of them at random (scaled by
    #     `density`).  Every occupied tile contributes equally regardless
    #     of how many cells it contains.
    # ------------------------------------------------------------------ #
    mass_flat      = mass.ravel()
    mass_threshold = np.percentile(
        mass_flat[mass_flat > 0], max(0, 100 - min_mass * 10)
    )
    valid_mask = mass_flat > mass_threshold
    valid_pts  = grid_pts[valid_mask]

    if len(valid_pts) == 0:
        raise RuntimeError("cl.stream(): no valid seed points. Lower min_mass.")

    # Build tile edges
    tile_x = np.linspace(x_min, x_max, n_tiles + 1)
    tile_y = np.linspace(y_min, y_max, n_tiles + 1)

    rng = np.random.default_rng(0)
    k   = max(1, int(round(seeds_per_tile * density)))  # seeds per occupied tile

    seeds = []
    for i in range(n_tiles):
        for j in range(n_tiles):
            # Grid points inside this tile AND above mass threshold
            in_tile = (
                (valid_pts[:, 0] >= tile_x[i]) & (valid_pts[:, 0] < tile_x[i + 1]) &
                (valid_pts[:, 1] >= tile_y[j]) & (valid_pts[:, 1] < tile_y[j + 1])
            )
            tile_pts = valid_pts[in_tile]
            if len(tile_pts) == 0:
                continue
            chosen = rng.choice(len(tile_pts), size=min(k, len(tile_pts)), replace=False)
            seeds.extend(tile_pts[chosen])

    seeds = np.array(seeds)
    if len(seeds) == 0:
        raise RuntimeError("cl.stream(): no seeds after tiling. Lower min_mass or raise n_tiles.")

    # KD-tree for manifold clipping
    cell_tree     = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt        = step_size if step_size is not None else cell_size * 0.8
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists > max_cell_dist))
            if np.any(dists > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=line_alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
        alpha=line_alpha,
    )

    return layer_path + layer_arrows
```

```python
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream5(
    dim,
    n_grid: int = 30,
    density: float = 1.0,
    color: str = "#1f1f1f",
    linewidth: float = 1.0,
    line_alpha: float = 0.6,
    arrow_size: float = 3,
    min_mass: float = 4.0,
    integration_steps: int = 60,
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.05,
    arrow_every: int = 12,
    seeds_per_tile: int = 1,
    n_tiles: int = 15,
    min_seed_dist_frac: float = 0.04,
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    Seeding strategy (two complementary passes):
      1. Tile grid  — divides the bounding box into n_tiles×n_tiles equal tiles;
                      each occupied tile gets seeds_per_tile seeds from the KDE
                      grid, giving uniform spatial coverage.
      2. Cell seeds — a random sub-sample of actual cell positions (jittered),
                      ensuring every part of the manifold that contains real
                      cells is seeded even if the grid tiles are too coarse.
    All candidate seeds are then deduplicated by a minimum-distance filter so
    no region ends up over-crowded.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : velocity interpolation grid resolution (default 30)
    density             : global seed count multiplier (default 1.0)
    color               : line and arrowhead colour (default "#1f1f1f")
    linewidth           : path line width (default 1.0)
    line_alpha          : opacity of lines and arrows (default 0.6)
    arrow_size          : arrowhead size in px (default 3)
    min_mass            : KDE floor — grid points below this percentile are
                          excluded from tile seeds (default 4.0)
    integration_steps   : max ODE steps per streamline (default 60)
    step_size           : ODE step size; auto = ~80 % of one grid cell
    max_cell_dist_frac  : clip streamline when it strays further than this
                          fraction of the data range from any real cell (0.05)
    arrow_every         : arrowhead every N path points (default 12)
    seeds_per_tile      : seeds per occupied tile (default 1)
    n_tiles             : tile grid resolution (default 15 → 15×15 tiles)
    min_seed_dist_frac  : minimum distance between any two seeds, as fraction
                          of the shorter data-range axis (default 0.04)
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names
    # ------------------------------------------------------------------ #
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    x  = frame[x_col].to_numpy().astype(np.float64)
    y  = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Regular grid + velocity interpolation
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    x_min, x_max = x.min() - x_range * 0.02, x.max() + x_range * 0.02
    y_min, y_max = y.min() - y_range * 0.02, y.max() + y_range * 0.02

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY   = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])

    bw_x = x_range / n_grid * 1.5
    bw_y = y_range / n_grid * 1.5

    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w  = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    speed   = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Dual seeding strategy
    # ------------------------------------------------------------------ #
    rng = np.random.default_rng(0)

    # --- pass A: tile-grid seeds (uniform spatial coverage) -------------
    mass_flat      = mass.ravel()
    mass_threshold = np.percentile(
        mass_flat[mass_flat > 0], max(0, 100 - min_mass * 10)
    )
    valid_mask = mass_flat > mass_threshold
    valid_pts  = grid_pts[valid_mask]

    if len(valid_pts) == 0:
        raise RuntimeError("cl.stream(): no valid seed points. Lower min_mass.")

    tile_x = np.linspace(x_min, x_max, n_tiles + 1)
    tile_y = np.linspace(y_min, y_max, n_tiles + 1)
    k      = max(1, int(round(seeds_per_tile * density)))

    tile_seeds = []
    for i in range(n_tiles):
        for j in range(n_tiles):
            in_tile = (
                (valid_pts[:, 0] >= tile_x[i]) & (valid_pts[:, 0] < tile_x[i + 1]) &
                (valid_pts[:, 1] >= tile_y[j]) & (valid_pts[:, 1] < tile_y[j + 1])
            )
            tile_pts = valid_pts[in_tile]
            if len(tile_pts) == 0:
                continue
            chosen = rng.choice(len(tile_pts), size=min(k, len(tile_pts)), replace=False)
            tile_seeds.extend(tile_pts[chosen])

    # --- pass B: jittered cell seeds (guarantees manifold coverage) -----
    # Sample ~1 seed per tile worth of cells, jittered by ±half a grid cell
    n_cell_seeds = max(10, n_tiles * n_tiles // 3)
    n_cell_seeds = min(n_cell_seeds, len(x))
    cell_idx     = rng.choice(len(x), size=n_cell_seeds, replace=False)
    jitter_x     = rng.uniform(-bw_x * 0.3, bw_x * 0.3, size=n_cell_seeds)
    jitter_y     = rng.uniform(-bw_y * 0.3, bw_y * 0.3, size=n_cell_seeds)
    cell_seeds   = np.column_stack([
        x[cell_idx] + jitter_x,
        y[cell_idx] + jitter_y,
    ])

    # --- merge both passes ----------------------------------------------
    all_seeds = np.vstack([tile_seeds, cell_seeds]) if tile_seeds else cell_seeds

    # Deduplicate by minimum inter-seed distance (order-independent shuffle
    # first so neither pass is systematically favoured)
    rng.shuffle(all_seeds)
    min_dist = min(x_range, y_range) * min_seed_dist_frac
    selected = [all_seeds[0]]
    for pt in all_seeds[1:]:
        if np.linalg.norm(np.array(selected) - pt, axis=1).min() >= min_dist:
            selected.append(pt)
    seeds = np.array(selected)

    # KD-tree for manifold clipping
    cell_tree     = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt        = step_size if step_size is not None else cell_size * 0.8
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists > max_cell_dist))
            if np.any(dists > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=line_alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
        alpha=line_alpha,
    )

    return layer_path + layer_arrows
```

```python
import numpy as np
import polars as pl
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream(
    dim,
    n_grid: int = 50,               # finer grid captures tight spirals
    density: float = 1.0,
    color: str = "#1f1f1f",
    linewidth: float = 0.8,
    line_alpha: float = 0.6,
    arrow_size: float = 3,
    min_mass: float = 4.0,
    integration_steps: int = 80,    # more steps → longer, fuller curves
    step_size: float | None = None,
    max_cell_dist_frac: float = 0.05,
    arrow_every: int = 12,
    seeds_per_tile: int = 1,
    n_tiles: int = 15,
    min_seed_dist_frac: float = 0.04,
    bw_scale: float = 3.5,          # wider bandwidth → smoother field
):
    """
    Compute and return LetsPlot layers (geom_path + geom_segment) showing
    RNA-velocity streamlines, similar to scVelo's stream plot.

    The key to scVelo-style smooth curvy streamlines is a wide KDE bandwidth
    (bw_scale ~3-4 grid cells) which smooths the velocity field before
    integration, producing sweeping curves rather than jagged paths.

    Parameters
    ----------
    dim                 : plot object returned by cl.umap(...)
    n_grid              : velocity interpolation grid resolution (default 50)
    density             : global seed count multiplier (default 1.0)
    color               : line and arrowhead colour (default "#1f1f1f")
    linewidth           : path line width (default 1.0)
    line_alpha          : opacity of lines and arrows (default 0.6)
    arrow_size          : arrowhead size in px (default 3)
    min_mass            : KDE floor for seeding (default 4.0)
    integration_steps   : max ODE steps per streamline (default 80)
    step_size           : ODE step size; auto = ~80 % of one grid cell
    max_cell_dist_frac  : clip streamline when it strays further than this
                          fraction of the data range from any real cell (0.05)
    arrow_every         : arrowhead every N path points (default 12)
    seeds_per_tile      : seeds per occupied tile (default 1)
    n_tiles             : tile grid resolution (default 15 → 15×15)
    min_seed_dist_frac  : minimum distance between seeds as fraction of the
                          shorter data-range axis (default 0.04)
    bw_scale            : KDE bandwidth in units of grid cells (default 3.5);
                          higher = smoother curves, lower = more local detail

    Tuning tips
    -----------
    Curves too jagged        → raise bw_scale (e.g. 4.0–5.0)
    Curves too smooth/blurry → lower bw_scale (e.g. 2.5)
    Spiral not captured      → raise n_grid (e.g. 60) or lower bw_scale slightly
    Lines leave data cloud   → lower max_cell_dist_frac (e.g. 0.03)
    Too few / many lines     → adjust n_tiles or min_seed_dist_frac
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names
    # ------------------------------------------------------------------ #
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    x  = frame[x_col].to_numpy().astype(np.float64)
    y  = frame[y_col].to_numpy().astype(np.float64)
    vx = frame[vx_col].to_numpy().astype(np.float64)
    vy = frame[vy_col].to_numpy().astype(np.float64)

    # ------------------------------------------------------------------ #
    # 2.  Regular grid + smooth velocity interpolation
    # ------------------------------------------------------------------ #
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    x_min, x_max = x.min() - x_range * 0.02, x.max() + x_range * 0.02
    y_min, y_max = y.min() - y_range * 0.02, y.max() + y_range * 0.02

    gx = np.linspace(x_min, x_max, n_grid)
    gy = np.linspace(y_min, y_max, n_grid)
    GX, GY   = np.meshgrid(gx, gy)
    grid_pts = np.column_stack([GX.ravel(), GY.ravel()])

    # Wide bandwidth: smooth over multiple grid cells so the integrated
    # field produces long sweeping arcs rather than noisy jitter
    bw_x = (x_range / n_grid) * bw_scale
    bw_y = (y_range / n_grid) * bw_scale

    def _kde(pts):
        dx = (pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        return np.exp(-0.5 * (dx**2 + dy**2)).sum(axis=1)

    mass = _kde(grid_pts).reshape(n_grid, n_grid)

    def _interp_vel(v_obs):
        dx = (grid_pts[:, 0:1] - x[np.newaxis, :]) / bw_x
        dy = (grid_pts[:, 1:2] - y[np.newaxis, :]) / bw_y
        w  = np.exp(-0.5 * (dx**2 + dy**2))
        return (w @ v_obs) / (w.sum(axis=1) + 1e-10)

    VX_grid = _interp_vel(vx).reshape(n_grid, n_grid)
    VY_grid = _interp_vel(vy).reshape(n_grid, n_grid)

    # Normalise to uniform speed so no region stalls
    speed   = np.sqrt(VX_grid**2 + VY_grid**2) + 1e-10
    VX_norm = VX_grid / speed
    VY_norm = VY_grid / speed

    # Zero out grid points with no cell support so paths don't drift into voids
    zero_mask = mass.reshape(n_grid, n_grid) < (mass.max() * 0.01)
    VX_norm[zero_mask] = 0.0
    VY_norm[zero_mask] = 0.0

    interp_vx = RegularGridInterpolator(
        (gy, gx), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (gy, gx), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [interp_vx(pt)[0], interp_vy(pt)[0]]

    # ------------------------------------------------------------------ #
    # 3.  Dual seeding: tile grid + jittered cell positions
    # ------------------------------------------------------------------ #
    rng = np.random.default_rng(0)

    # Pass A: tile-grid seeds
    mass_flat      = mass.ravel()
    mass_threshold = np.percentile(
        mass_flat[mass_flat > 0], max(0, 100 - min_mass * 10)
    )
    valid_mask = mass_flat > mass_threshold
    valid_pts  = grid_pts[valid_mask]

    if len(valid_pts) == 0:
        raise RuntimeError("cl.stream(): no valid seed points. Lower min_mass.")

    tile_x = np.linspace(x_min, x_max, n_tiles + 1)
    tile_y = np.linspace(y_min, y_max, n_tiles + 1)
    k      = max(1, int(round(seeds_per_tile * density)))

    tile_seeds = []
    for i in range(n_tiles):
        for j in range(n_tiles):
            in_tile = (
                (valid_pts[:, 0] >= tile_x[i]) & (valid_pts[:, 0] < tile_x[i + 1]) &
                (valid_pts[:, 1] >= tile_y[j]) & (valid_pts[:, 1] < tile_y[j + 1])
            )
            tile_pts = valid_pts[in_tile]
            if len(tile_pts) == 0:
                continue
            chosen = rng.choice(len(tile_pts), size=min(k, len(tile_pts)), replace=False)
            tile_seeds.extend(tile_pts[chosen])

    # Pass B: jittered cell seeds — guarantees manifold coverage
    n_cell_seeds = max(10, n_tiles * n_tiles // 3)
    n_cell_seeds = min(n_cell_seeds, len(x))
    cell_idx     = rng.choice(len(x), size=n_cell_seeds, replace=False)
    jitter_x     = rng.uniform(-bw_x * 0.3, bw_x * 0.3, size=n_cell_seeds)
    jitter_y     = rng.uniform(-bw_y * 0.3, bw_y * 0.3, size=n_cell_seeds)
    cell_seeds   = np.column_stack([
        x[cell_idx] + jitter_x,
        y[cell_idx] + jitter_y,
    ])

    # Merge, shuffle, deduplicate by min distance
    all_seeds = np.vstack([tile_seeds, cell_seeds]) if tile_seeds else cell_seeds
    rng.shuffle(all_seeds)

    min_dist = min(x_range, y_range) * min_seed_dist_frac
    selected = [all_seeds[0]]
    for pt in all_seeds[1:]:
        if np.linalg.norm(np.array(selected) - pt, axis=1).min() >= min_dist:
            selected.append(pt)
    seeds = np.array(selected)

    # KD-tree for manifold clipping
    cell_tree     = cKDTree(np.column_stack([x, y]))
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 4.  Integrate streamlines forward
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt        = step_size if step_size is not None else cell_size * 0.8
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists > max_cell_dist))
            if np.any(dists > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    # ------------------------------------------------------------------ #
    # 5.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, max_cell_dist_frac, or increasing n_grid."
        )

    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id"),
        color=color,
        size=linewidth,
        alpha=line_alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth * 1.5,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
        alpha=line_alpha,
    )

    return layer_path + layer_arrows
```

```python
import numpy as np
import polars as pl
from scipy.stats import norm as normal
from sklearn.neighbors import NearestNeighbors
from scipy.interpolate import RegularGridInterpolator
from scipy.integrate import solve_ivp
from scipy.spatial import cKDTree
from lets_plot import geom_path, geom_segment, arrow, aes
from cellestial import retrieve, get_mapping


def stream(
    dim,
    density: float = 2.0,
    smooth: float = 0.5,
    min_mass: float = 1.0,
    cutoff_perc: float = 5.0,
    n_neighbors: int | None = None,
    color: str = "#1f1f1f",
    linewidth: float = 1.0,
    line_alpha: float = 0.6,
    arrow_size: float = 3,
    integration_steps: int = 80,
    arrow_every: int = 12,
    max_cell_dist_frac: float = 0.05,
):
    """
    Stream plot of RNA velocities, faithfully ported from scVelo's
    compute_velocity_on_grid + velocity_embedding_stream.

    Parameters
    ----------
    dim              : plot object returned by cl.umap(...)
    density          : controls grid resolution — int(50 * density) points per
                       axis (default 2 → 100×100 grid, same as scVelo default)
    smooth           : Gaussian kernel scale = mean_grid_spacing * smooth
                       (default 0.5, same as scVelo)
    min_mass         : velocity-magnitude cutoff: 10^(min_mass - 6)
                       range 0 (all) to 5 (large velocities only), default 1
    cutoff_perc      : additionally mask grid points where velocity length is
                       below this percentile (default 5, same as scVelo)
    n_neighbors      : neighbours for velocity interpolation
                       (default n_obs // 50, same as scVelo)
    color            : streamline colour (default "#1f1f1f")
    linewidth        : base line width, scaled per-line by velocity magnitude
                       (default 1.0)
    line_alpha       : opacity (default 0.6)
    arrow_size       : arrowhead size in px (default 3)
    integration_steps: ODE steps per streamline (default 80)
    arrow_every      : arrowhead every N path points (default 12)
    max_cell_dist_frac: clip streamline when it strays further than this
                        fraction of the shorter data-range axis (default 0.05)

    Tuning tips (mirrors scVelo)
    ----------------------------
    More / denser lines    → raise density (e.g. 3)
    Smoother velocity field → raise smooth (e.g. 1.0)
    Only fast regions      → raise min_mass (e.g. 3)
    Remove short stubs     → raise cutoff_perc (e.g. 10)
    """
    # ------------------------------------------------------------------ #
    # 1.  Pull data and column names
    # ------------------------------------------------------------------ #
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    X_emb = frame.select([x_col, y_col]).to_numpy().astype(np.float64)
    V_emb = frame.select([vx_col, vy_col]).to_numpy().astype(np.float64)

    # Remove invalid cells (NaN / Inf)
    idx_valid = np.isfinite(X_emb.sum(1) + V_emb.sum(1))
    X_emb = X_emb[idx_valid]
    V_emb = V_emb[idx_valid]

    n_obs = len(X_emb)

    # ------------------------------------------------------------------ #
    # 2.  Build grid — scVelo: int(50 * density) points per axis
    # ------------------------------------------------------------------ #
    n_grid = int(50 * density)

    grs = []
    for dim_i in range(2):
        m, M = X_emb[:, dim_i].min(), X_emb[:, dim_i].max()
        m = m - 0.01 * abs(M - m)
        M = M + 0.01 * abs(M - m)
        grs.append(np.linspace(m, M, n_grid))

    meshes = np.meshgrid(*grs)
    X_grid_flat = np.vstack([g.flat for g in meshes]).T   # (n_grid^2, 2)

    # ------------------------------------------------------------------ #
    # 3.  Velocity interpolation — scVelo: kNN + Gaussian weight
    # ------------------------------------------------------------------ #
    if n_neighbors is None:
        n_neighbors = max(1, int(n_obs / 50))

    nn = NearestNeighbors(n_neighbors=n_neighbors, n_jobs=-1)
    nn.fit(X_emb)
    dists, neighs = nn.kneighbors(X_grid_flat)

    # scale = mean grid spacing * smooth  (exactly scVelo)
    scale  = np.mean([(g[1] - g[0]) for g in grs]) * smooth
    weight = normal.pdf(x=dists, scale=scale)   # (n_grid^2, k)
    p_mass = weight.sum(1)                       # (n_grid^2,)

    V_grid_flat = (V_emb[neighs] * weight[:, :, None]).sum(1)  # (n_grid^2, 2)
    V_grid_flat /= np.maximum(1, p_mass)[:, None]

    # ------------------------------------------------------------------ #
    # 4.  Masking — scVelo adjust_for_stream=True logic
    # ------------------------------------------------------------------ #
    # Reshape to 2D grids
    X_grid = np.stack([grs[0], grs[1]])         # (2, n_grid)
    V_grid = V_grid_flat.T.reshape(2, n_grid, n_grid)   # (2, n_grid, n_grid)

    mass    = np.sqrt((V_grid**2).sum(0))        # velocity magnitude on grid
    min_mass_val = 10 ** (min_mass - 6)          # scVelo: 10^(min_mass - 6)
    min_mass_val = np.clip(min_mass_val, None, mass.max() * 0.9)

    # Length mask: mean absolute velocity of neighbours per grid point
    length = np.sum(
        np.mean(np.abs(V_emb[neighs]), axis=1), axis=1
    ).T.reshape(n_grid, n_grid)

    cutoff  = mass < min_mass_val
    cutoff |= length < np.percentile(length, cutoff_perc)

    # NaN-out masked cells — streamlines that enter these regions will stall
    VX = V_grid[0].copy();  VX[cutoff] = np.nan
    VY = V_grid[1].copy();  VY[cutoff] = np.nan

    # Per-line speed for linewidth scaling (scVelo scales lw by 2*len/max_len)
    lengths_grid = np.sqrt(
        np.nan_to_num(VX)**2 + np.nan_to_num(VY)**2
    )
    max_len = lengths_grid[~np.isnan(VX)].max() if (~np.isnan(VX)).any() else 1.0

    # ------------------------------------------------------------------ #
    # 5.  Build interpolators for ODE integration
    #     Replace NaN with 0 so the integrator doesn't blow up;
    #     the KD-tree clipping handles manifold boundaries.
    # ------------------------------------------------------------------ #
    VX_safe = np.nan_to_num(VX, nan=0.0)
    VY_safe = np.nan_to_num(VY, nan=0.0)

    interp_vx = RegularGridInterpolator(
        (grs[1], grs[0]), VX_safe, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (grs[1], grs[0]), VY_safe, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_lw = RegularGridInterpolator(
        (grs[1], grs[0]), lengths_grid, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t, state):
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [float(interp_vx(pt)), float(interp_vy(pt))]

    # ------------------------------------------------------------------ #
    # 6.  Seed points: non-masked grid points only
    # ------------------------------------------------------------------ #
    valid_mask = ~cutoff.ravel()
    gx_flat = X_grid_flat[:, 0]
    gy_flat = X_grid_flat[:, 1]
    seeds   = np.column_stack([gx_flat[valid_mask], gy_flat[valid_mask]])

    # Subsample so we don't integrate from every grid point
    rng      = np.random.default_rng(0)
    n_seeds  = max(50, int(len(seeds) * min(density / 4.0, 1.0)))
    idx      = rng.choice(len(seeds), size=min(n_seeds, len(seeds)), replace=False)
    seeds    = seeds[idx]

    # KD-tree for manifold clipping
    x_range      = X_emb[:, 0].max() - X_emb[:, 0].min()
    y_range      = X_emb[:, 1].max() - X_emb[:, 1].min()
    cell_tree     = cKDTree(X_emb)
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # ------------------------------------------------------------------ #
    # 7.  Integrate streamlines
    # ------------------------------------------------------------------ #
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt        = cell_size * 0.8
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        # Clip at first point that leaves the manifold
        dists_path, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists_path > max_cell_dist))
            if np.any(dists_path > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue
        xs = xs[:valid_end]
        ys = ys[:valid_end]

        # scVelo-style per-line linewidth: 2 * local_speed / max_speed
        mid     = len(xs) // 2
        pt_mid  = np.array([[ys[mid], xs[mid]]])
        lw_here = float(interp_lw(pt_mid))
        lw_line = linewidth * 2 * lw_here / (max_len + 1e-10)
        lw_line = float(np.clip(lw_line, linewidth * 0.3, linewidth * 3.0))

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({
                "x": float(xi), "y": float(yi),
                "line_id": line_id, "lw": lw_line,
            })

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
                "lw":   lw_line,
            })

    # ------------------------------------------------------------------ #
    # 8.  Build LetsPlot layers
    # ------------------------------------------------------------------ #
    if not path_rows:
        raise RuntimeError(
            "cl.stream(): no streamlines generated. "
            "Try lowering min_mass, raising density, or lowering cutoff_perc."
        )

    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id", size="lw"),
        color=color,
        alpha=line_alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
        alpha=line_alpha,
    )

    return layer_path + layer_arrows
```

```python
from __future__ import annotations

import numpy as np
import polars as pl
from scipy.integrate import solve_ivp
from scipy.interpolate import RegularGridInterpolator
from scipy.spatial import cKDTree
from scipy.stats import norm as normal
from sklearn.neighbors import NearestNeighbors

from lets_plot import aes, geom_path, geom_segment, arrow
from lets_plot.plot.core import FeatureSpec

from cellestial import get_mapping, retrieve


def stream(
    dim: FeatureSpec,
    *,
    density: float = 2.0,
    smooth: float = 0.5,
    min_mass: float = 1.0,
    cutoff_perc: float = 5.0,
    n_neighbors: int | None = None,
    integration_steps: int = 80,
    max_cell_dist_frac: float = 0.05,
    color: str = "#1f1f1f",
    linewidth: float = 1.0,
    alpha: float = 0.6,
    arrow_size: float = 3,
    arrow_every: int = 12,
) -> FeatureSpec:
    """
    Stream plot of RNA velocities overlaid on a dimensionality reduction plot.

    Parameters
    ----------
    dim : FeatureSpec
        The plot object returned by ``cl.umap()`` or similar.
    density : float, default=2.0
        Controls grid resolution — ``int(50 * density)`` points per axis.
        Higher values produce more streamlines.
    smooth : float, default=0.5
        Gaussian kernel scale multiplier applied to the mean grid spacing.
        Higher values smooth the velocity field more.
    min_mass : float, default=1.0
        Velocity-magnitude cutoff: ``10 ** (min_mass - 6)``.
        Range: 0 (all velocities shown) to 5 (large velocities only).
    cutoff_perc : float, default=5.0
        Mask grid points where neighbour velocity length falls below this
        percentile (0–100).
    n_neighbors : int | None, default=None
        Number of neighbours for velocity interpolation.
        Defaults to ``max(1, n_obs // 50)``.
    integration_steps : int, default=80
        Maximum number of ODE integration steps per streamline.
    max_cell_dist_frac : float, default=0.05
        Streamlines are clipped when they stray further than this fraction
        of the shorter data-range axis from any real cell.
    color : str, default='#1f1f1f'
        Colour of the streamlines and arrowheads.

        **Accepts:**

        - hex code e.g. ``'#f1f1f1'``
        - color name (https://lets-plot.org/python/pages/named_colors.html).
        - RGB/RGBA e.g. ``'rgb(0, 0, 255)'``, ``'rgba(0, 0, 255, 0.5)'``.

    linewidth : float, default=1.0
        Base line width, scaled per streamline by local velocity magnitude.
    alpha : float, default=0.6
        Opacity of the streamlines and arrowheads.
    arrow_size : float, default=3
        Arrowhead size in px.
    arrow_every : int, default=12
        Place an arrowhead every ``arrow_every`` path points.

    Returns
    -------
    FeatureSpec
        A LetsPlot layer (``geom_path`` + ``geom_segment``) that can be added
        to an existing plot with ``+``.

    Examples
    --------
    Basic stream overlay.

    .. jupyter-execute::

        import scanpy as sc
        import cellestial as cl

        data = sc.read_h5ad("data/pancreas.h5ad")

        dim = cl.umap(data, key="clusters_coarse", axis_type="arrow")
        dim + cl.stream(dim)

    Denser, smoother streamlines.

    .. jupyter-execute::

        dim + cl.stream(dim, density=3.0, smooth=1.0, min_mass=2.0)
    """
    # HANDLE: retrieve frame and column names from the plot object
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    X_emb = frame.select([x_col, y_col]).to_numpy().astype(np.float64)
    V_emb = frame.select([vx_col, vy_col]).to_numpy().astype(np.float64)

    # HANDLE: remove invalid cells
    idx_valid = np.isfinite(X_emb.sum(1) + V_emb.sum(1))
    X_emb = X_emb[idx_valid]
    V_emb = V_emb[idx_valid]

    n_obs = len(X_emb)

    # BUILD: velocity grid
    n_grid = int(50 * density)

    grs = []
    for dim_i in range(2):
        m, M = X_emb[:, dim_i].min(), X_emb[:, dim_i].max()
        m = m - 0.01 * abs(M - m)
        M = M + 0.01 * abs(M - m)
        grs.append(np.linspace(m, M, n_grid))

    meshes     = np.meshgrid(*grs)
    X_grid_flat = np.vstack([g.flat for g in meshes]).T

    # BUILD: kNN velocity interpolation
    if n_neighbors is None:
        n_neighbors = max(1, int(n_obs / 50))

    nn = NearestNeighbors(n_neighbors=n_neighbors, n_jobs=-1)
    nn.fit(X_emb)
    dists, neighs = nn.kneighbors(X_grid_flat)

    scale       = np.mean([(g[1] - g[0]) for g in grs]) * smooth
    weight      = normal.pdf(x=dists, scale=scale)
    p_mass      = weight.sum(1)

    V_grid_flat  = (V_emb[neighs] * weight[:, :, None]).sum(1)
    V_grid_flat /= np.maximum(1, p_mass)[:, None]

    # HANDLE: masking — velocity magnitude + length percentile cutoff
    X_grid = np.stack([grs[0], grs[1]])
    V_grid = V_grid_flat.T.reshape(2, n_grid, n_grid)

    mass         = np.sqrt((V_grid**2).sum(0))
    min_mass_val = 10 ** (min_mass - 6)
    min_mass_val = np.clip(min_mass_val, None, mass.max() * 0.9)

    length = np.sum(
        np.mean(np.abs(V_emb[neighs]), axis=1), axis=1
    ).T.reshape(n_grid, n_grid)

    cutoff  = mass < min_mass_val
    cutoff |= length < np.percentile(length, cutoff_perc)

    VX = V_grid[0].copy(); VX[cutoff] = np.nan
    VY = V_grid[1].copy(); VY[cutoff] = np.nan

    lengths_grid = np.sqrt(np.nan_to_num(VX) ** 2 + np.nan_to_num(VY) ** 2)
    max_len      = lengths_grid[~np.isnan(VX)].max() if (~np.isnan(VX)).any() else 1.0

    # BUILD: interpolators for ODE integration
    VX_safe = np.nan_to_num(VX, nan=0.0)
    VY_safe = np.nan_to_num(VY, nan=0.0)

    interp_vx = RegularGridInterpolator(
        (grs[1], grs[0]), VX_safe, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (grs[1], grs[0]), VY_safe, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_lw = RegularGridInterpolator(
        (grs[1], grs[0]), lengths_grid, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t: float, state: list[float]) -> list[float]:
        pt = np.array([[state[1], state[0]]])
        return [float(interp_vx(pt)), float(interp_vy(pt))]

    # HANDLE: seed points from non-masked grid cells
    valid_mask = ~cutoff.ravel()
    seeds      = X_grid_flat[valid_mask]

    rng     = np.random.default_rng(0)
    n_seeds = max(50, int(len(seeds) * min(density / 4.0, 1.0)))
    idx     = rng.choice(len(seeds), size=min(n_seeds, len(seeds)), replace=False)
    seeds   = seeds[idx]

    x_range       = X_emb[:, 0].max() - X_emb[:, 0].min()
    y_range       = X_emb[:, 1].max() - X_emb[:, 1].min()
    cell_tree     = cKDTree(X_emb)
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # BUILD: streamlines via ODE integration
    cell_size = ((x_range + y_range) / 2) / n_grid
    dt        = cell_size * 0.8
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists_path, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists_path > max_cell_dist))
            if np.any(dists_path > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue

        xs = xs[:valid_end]
        ys = ys[:valid_end]

        mid     = len(xs) // 2
        lw_here = float(interp_lw(np.array([[ys[mid], xs[mid]]])))
        lw_line = float(np.clip(
            linewidth * 2 * lw_here / (max_len + 1e-10),
            linewidth * 0.3,
            linewidth * 3.0,
        ))

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id, "lw": lw_line})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    if not path_rows:
        msg = (
            "stream(): no streamlines generated. "
            "Try lowering min_mass, raising density, or lowering cutoff_perc."
        )
        raise RuntimeError(msg)

    # BUILD: LetsPlot layers
    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id", size="lw"),
        color=color,
        alpha=alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth,
        alpha=alpha,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
```

```python
from __future__ import annotations

import numpy as np
import polars as pl
from scipy.integrate import solve_ivp
from scipy.interpolate import RegularGridInterpolator
from scipy.spatial import cKDTree
from scipy.stats import norm as normal
from sklearn.neighbors import NearestNeighbors

from lets_plot import aes, arrow, geom_path, geom_segment
from lets_plot.plot.core import FeatureSpec

from cellestial import get_mapping, retrieve


def stream(
    dim: FeatureSpec,
    *,
    density: float = 2.0,
    smooth: float = 0.5,
    min_mass: float = 1.0,
    cutoff_perc: float = 5.0,
    n_neighbors: int | None = None,
    integration_steps: int = 80,
    max_cell_dist_frac: float = 0.05,
    color: str = "#1f1f1f",
    linewidth: float = 1.0,
    alpha: float = 0.6,
    arrow_size: float = 3,
    arrow_every: int = 12,
) -> FeatureSpec:
    """
    Stream plot of RNA velocities overlaid on a dimensionality reduction plot.

    Parameters
    ----------
    dim : FeatureSpec
        The plot object returned by ``cl.umap()`` or similar.
    density : float, default=2.0
        Controls grid resolution — ``int(50 * density)`` points per axis.
        Higher values produce more streamlines.
    smooth : float, default=0.5
        Gaussian kernel scale multiplier applied to the mean grid spacing.
        Higher values smooth the velocity field more.
    min_mass : float, default=1.0
        Velocity-magnitude cutoff: ``10 ** (min_mass - 6)``.
        Range: 0 (all velocities shown) to 5 (large velocities only).
    cutoff_perc : float, default=5.0
        Mask grid points where neighbour velocity length falls below this
        percentile (0–100).
    n_neighbors : int | None, default=None
        Number of neighbours for velocity interpolation.
        Defaults to ``max(1, n_obs // 50)``.
    integration_steps : int, default=80
        Maximum number of ODE integration steps per streamline.
    max_cell_dist_frac : float, default=0.05
        Streamlines are clipped when they stray further than this fraction
        of the shorter data-range axis from any real cell.
    color : str, default='#1f1f1f'
        Colour of the streamlines and arrowheads.

        **Accepts:**

        - hex code e.g. ``'#f1f1f1'``
        - color name (https://lets-plot.org/python/pages/named_colors.html).
        - RGB/RGBA e.g. ``'rgb(0, 0, 255)'``, ``'rgba(0, 0, 255, 0.5)'``.

    linewidth : float, default=1.0
        Base line width, scaled per streamline by local velocity magnitude.
    alpha : float, default=0.6
        Opacity of the streamlines and arrowheads.
    arrow_size : float, default=3
        Arrowhead size in px.
    arrow_every : int, default=12
        Place an arrowhead every ``arrow_every`` path points.

    Returns
    -------
    FeatureSpec
        A LetsPlot layer (``geom_path`` + ``geom_segment``) that can be added
        to an existing plot with ``+``.

    Examples
    --------
    Basic stream overlay.

    .. jupyter-execute::

        import scanpy as sc
        import cellestial as cl

        data = sc.read_h5ad("data/pancreas.h5ad")

        dim = cl.umap(data, key="clusters_coarse", axis_type="arrow")
        dim + cl.stream(dim)

    Denser, smoother streamlines.

    .. jupyter-execute::

        dim + cl.stream(dim, density=3.0, smooth=1.0, min_mass=2.0)
    """
    # HANDLE: retrieve frame and column names from the plot object
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    X_emb = frame.select([x_col, y_col]).to_numpy().astype(np.float64)
    V_emb = frame.select([vx_col, vy_col]).to_numpy().astype(np.float64)

    # HANDLE: remove invalid cells
    idx_valid = np.isfinite(X_emb.sum(1) + V_emb.sum(1))
    X_emb = X_emb[idx_valid]
    V_emb = V_emb[idx_valid]

    n_obs = len(X_emb)

    # BUILD: velocity grid
    n_grid = int(50 * density)

    grs = []
    for dim_i in range(2):
        m, M = X_emb[:, dim_i].min(), X_emb[:, dim_i].max()
        m = m - 0.01 * abs(M - m)
        M = M + 0.01 * abs(M - m)
        grs.append(np.linspace(m, M, n_grid))

    meshes      = np.meshgrid(*grs)
    X_grid_flat = np.vstack([g.flat for g in meshes]).T

    # BUILD: kNN velocity interpolation
    if n_neighbors is None:
        n_neighbors = max(1, int(n_obs / 50))

    nn = NearestNeighbors(n_neighbors=n_neighbors, n_jobs=-1)
    nn.fit(X_emb)
    dists, neighs = nn.kneighbors(X_grid_flat)

    scale        = np.mean([(g[1] - g[0]) for g in grs]) * smooth
    weight       = normal.pdf(x=dists, scale=scale)
    p_mass       = weight.sum(1)

    V_grid_flat  = (V_emb[neighs] * weight[:, :, None]).sum(1)
    V_grid_flat /= np.maximum(1, p_mass)[:, None]

    # HANDLE: masking — velocity magnitude + length percentile cutoff
    V_grid = V_grid_flat.T.reshape(2, n_grid, n_grid)

    mass         = np.sqrt((V_grid**2).sum(0))
    min_mass_val = 10 ** (min_mass - 6)
    min_mass_val = np.clip(min_mass_val, None, mass.max() * 0.9)

    length = np.sum(
        np.mean(np.abs(V_emb[neighs]), axis=1), axis=1
    ).T.reshape(n_grid, n_grid)

    cutoff  = mass < min_mass_val
    cutoff |= length < np.percentile(length, cutoff_perc)

    VX = V_grid[0].copy()
    VY = V_grid[1].copy()

    # Normalise to unit speed for integration so dt is embedding-scale,
    # not velocity-scale. This is essential when velocities are very small
    # (e.g. ~0.05) relative to the embedding range (~20 units): without
    # normalisation the ODE covers < 1 grid cell in 80 steps and all paths
    # are clipped. Linewidth is still scaled by the original magnitude.
    raw_speed    = np.sqrt(VX**2 + VY**2) + 1e-10
    lengths_grid = raw_speed.copy()           # keep original for lw scaling
    VX_norm      = VX / raw_speed
    VY_norm      = VY / raw_speed
    VX_norm[cutoff] = 0.0
    VY_norm[cutoff] = 0.0

    max_len = lengths_grid[~cutoff].max() if (~cutoff).any() else 1.0

    # BUILD: interpolators for ODE integration
    interp_vx = RegularGridInterpolator(
        (grs[1], grs[0]), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (grs[1], grs[0]), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_lw = RegularGridInterpolator(
        (grs[1], grs[0]), lengths_grid, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t: float, state: list[float]) -> list[float]:
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [float(interp_vx(pt)), float(interp_vy(pt))]

    # HANDLE: seed points from non-masked grid cells
    valid_mask = ~cutoff.ravel()
    seeds      = X_grid_flat[valid_mask]

    rng     = np.random.default_rng(0)
    n_seeds = max(50, int(len(seeds) * min(density / 4.0, 1.0)))
    idx     = rng.choice(len(seeds), size=min(n_seeds, len(seeds)), replace=False)
    seeds   = seeds[idx]

    x_range       = X_emb[:, 0].max() - X_emb[:, 0].min()
    y_range       = X_emb[:, 1].max() - X_emb[:, 1].min()
    cell_tree     = cKDTree(X_emb)
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # BUILD: streamlines via ODE integration
    # dt is a fraction of one grid cell in embedding space — independent of
    # velocity magnitude because we integrate the normalised field above
    grid_cell     = np.mean([(g[1] - g[0]) for g in grs])
    dt            = grid_cell * 0.5
    t_span        = (0.0, dt * integration_steps)
    t_eval        = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists_path, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists_path > max_cell_dist))
            if np.any(dists_path > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue

        xs = xs[:valid_end]
        ys = ys[:valid_end]

        mid     = len(xs) // 2
        lw_here = float(interp_lw(np.array([[ys[mid], xs[mid]]])))
        lw_line = float(np.clip(
            linewidth * 2 * lw_here / (max_len + 1e-10),
            linewidth * 0.3,
            linewidth * 3.0,
        ))

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id, "lw": lw_line})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    if not path_rows:
        msg = (
            "stream(): no streamlines generated. "
            "Try lowering min_mass, raising density, or lowering cutoff_perc."
        )
        raise RuntimeError(msg)

    # BUILD: LetsPlot layers
    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id", size="lw"),
        color=color,
        alpha=alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth,
        alpha=alpha,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
```

```python
from __future__ import annotations

import numpy as np
import polars as pl
from scipy.integrate import solve_ivp
from scipy.interpolate import RegularGridInterpolator
from scipy.spatial import cKDTree
from sklearn.neighbors import NearestNeighbors

from lets_plot import aes, arrow, geom_path, geom_segment
from lets_plot.plot.core import FeatureSpec

from cellestial import get_mapping, retrieve


def stream(
    dim: FeatureSpec,
    *,
    density: float = 2.0,
    smooth: float = 0.5,
    min_mass: float = 1.0,
    cutoff_perc: float = 5.0,
    n_neighbors: int | None = None,
    integration_steps: int = 80,
    max_cell_dist_frac: float = 0.05,
    color: str = "#1f1f1f",
    linewidth: float = 1.0,
    alpha: float = 0.6,
    arrow_size: float = 3,
    arrow_every: int = 12,
) -> FeatureSpec:
    """
    Stream plot of RNA velocities overlaid on a dimensionality reduction plot.

    Parameters
    ----------
    dim : FeatureSpec
        The plot object returned by ``cl.umap()`` or similar.
    density : float, default=2.0
        Controls grid resolution — ``int(50 * density)`` points per axis.
        Higher values produce more streamlines.
    smooth : float, default=0.5
        Gaussian kernel scale as a fraction of the embedding range.
        Higher values smooth the velocity field more.
    min_mass : float, default=1.0
        Velocity-magnitude cutoff: ``10 ** (min_mass - 6)``.
        Range: 0 (all velocities shown) to 5 (large velocities only).
    cutoff_perc : float, default=5.0
        Mask grid points where neighbour velocity length falls below this
        percentile (0–100).
    n_neighbors : int | None, default=None
        Number of neighbours for velocity interpolation.
        Defaults to ``max(1, n_obs // 50)``.
    integration_steps : int, default=80
        Maximum number of ODE integration steps per streamline.
    max_cell_dist_frac : float, default=0.05
        Streamlines are clipped when they stray further than this fraction
        of the shorter data-range axis from any real cell.
    color : str, default='#1f1f1f'
        Colour of the streamlines and arrowheads.

        **Accepts:**

        - hex code e.g. ``'#f1f1f1'``
        - color name (https://lets-plot.org/python/pages/named_colors.html).
        - RGB/RGBA e.g. ``'rgb(0, 0, 255)'``, ``'rgba(0, 0, 255, 0.5)'``.

    linewidth : float, default=1.0
        Base line width, scaled per streamline by local velocity magnitude.
    alpha : float, default=0.6
        Opacity of the streamlines and arrowheads.
    arrow_size : float, default=3
        Arrowhead size in px.
    arrow_every : int, default=12
        Place an arrowhead every ``arrow_every`` path points.

    Returns
    -------
    FeatureSpec
        A LetsPlot layer (``geom_path`` + ``geom_segment``) that can be added
        to an existing plot with ``+``.

    Examples
    --------
    Basic stream overlay.

    .. jupyter-execute::

        import scanpy as sc
        import cellestial as cl

        data = sc.read_h5ad("data/pancreas.h5ad")

        dim = cl.umap(data, key="clusters_coarse", axis_type="arrow")
        dim + cl.stream(dim)

    Denser, smoother streamlines.

    .. jupyter-execute::

        dim + cl.stream(dim, density=3.0, smooth=1.0, min_mass=2.0)
    """
    # HANDLE: retrieve frame and column names from the plot object
    frame   = retrieve(dim)
    mapping = get_mapping(dim)

    x_col  = mapping["x"]
    y_col  = mapping["y"]
    vx_col = f"VELOCITY_{x_col.replace('X_', '')}"
    vy_col = f"VELOCITY_{y_col.replace('X_', '')}"

    X_emb = frame.select([x_col, y_col]).to_numpy().astype(np.float64)
    V_emb = frame.select([vx_col, vy_col]).to_numpy().astype(np.float64)

    # HANDLE: remove invalid cells
    idx_valid = np.isfinite(X_emb.sum(1) + V_emb.sum(1))
    X_emb = X_emb[idx_valid]
    V_emb = V_emb[idx_valid]

    n_obs = len(X_emb)

    # BUILD: velocity grid
    n_grid = int(50 * density)

    grs = []
    for dim_i in range(2):
        m, M = X_emb[:, dim_i].min(), X_emb[:, dim_i].max()
        m = m - 0.01 * abs(M - m)
        M = M + 0.01 * abs(M - m)
        grs.append(np.linspace(m, M, n_grid))

    meshes      = np.meshgrid(*grs)
    X_grid_flat = np.vstack([g.flat for g in meshes]).T

    # BUILD: kNN velocity interpolation
    # Scale bandwidth to the embedding range, not grid spacing.
    # grid spacing ~0.11 with scale=0.5 gives kernel width ~0.055 which is
    # far too narrow for an embedding spanning ~22 units — almost all weights
    # collapse to zero and the velocity field vanishes.
    # Using a fraction of the mean embedding range keeps the kernel sensible
    # regardless of the embedding scale.
    if n_neighbors is None:
        n_neighbors = max(1, int(n_obs / 50))

    nn = NearestNeighbors(n_neighbors=n_neighbors, n_jobs=-1)
    nn.fit(X_emb)
    dists, neighs = nn.kneighbors(X_grid_flat)

    emb_range = np.mean([g[-1] - g[0] for g in grs])
    scale     = emb_range * smooth / n_grid   # fraction of embedding range per grid cell

    # Gaussian weights based on distance to neighbours
    weight       = np.exp(-0.5 * (dists / scale) ** 2)
    p_mass       = weight.sum(1)

    V_grid_flat  = (V_emb[neighs] * weight[:, :, None]).sum(1)
    V_grid_flat /= np.maximum(p_mass, 1e-10)[:, None]

    # HANDLE: masking — velocity magnitude + length percentile cutoff
    V_grid = V_grid_flat.T.reshape(2, n_grid, n_grid)

    mass         = np.sqrt((V_grid**2).sum(0))
    min_mass_val = 10 ** (min_mass - 6)
    min_mass_val = np.clip(min_mass_val, None, mass.max() * 0.9)

    length = np.sum(
        np.mean(np.abs(V_emb[neighs]), axis=1), axis=1
    ).T.reshape(n_grid, n_grid)

    cutoff  = mass < min_mass_val
    cutoff |= length < np.percentile(length, cutoff_perc)

    VX = V_grid[0].copy()
    VY = V_grid[1].copy()

    # Normalise to unit speed for integration so dt is in embedding-space
    # units regardless of velocity magnitude
    raw_speed    = np.sqrt(VX**2 + VY**2) + 1e-10
    lengths_grid = raw_speed.copy()
    VX_norm      = np.where(cutoff, 0.0, VX / raw_speed)
    VY_norm      = np.where(cutoff, 0.0, VY / raw_speed)

    max_len = lengths_grid[~cutoff].max() if (~cutoff).any() else 1.0

    # BUILD: interpolators for ODE integration
    interp_vx = RegularGridInterpolator(
        (grs[1], grs[0]), VX_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_vy = RegularGridInterpolator(
        (grs[1], grs[0]), VY_norm, method="linear", bounds_error=False, fill_value=0.0
    )
    interp_lw = RegularGridInterpolator(
        (grs[1], grs[0]), lengths_grid, method="linear", bounds_error=False, fill_value=0.0
    )

    def _field(t: float, state: list[float]) -> list[float]:
        pt = np.array([[state[1], state[0]]])   # (y, x)
        return [float(interp_vx(pt)), float(interp_vy(pt))]

    # HANDLE: seed points from non-masked grid cells
    valid_mask = ~cutoff.ravel()
    seeds      = X_grid_flat[valid_mask]

    if len(seeds) == 0:
        msg = (
            "stream(): no valid seed points found after masking. "
            "Try lowering min_mass or cutoff_perc."
        )
        raise RuntimeError(msg)

    rng     = np.random.default_rng(0)
    n_seeds = max(50, int(len(seeds) * min(density / 4.0, 1.0)))
    idx     = rng.choice(len(seeds), size=min(n_seeds, len(seeds)), replace=False)
    seeds   = seeds[idx]

    x_range       = X_emb[:, 0].max() - X_emb[:, 0].min()
    y_range       = X_emb[:, 1].max() - X_emb[:, 1].min()
    cell_tree     = cKDTree(X_emb)
    max_cell_dist = min(x_range, y_range) * max_cell_dist_frac

    # BUILD: streamlines via ODE integration
    # dt is a fraction of one grid cell in embedding space
    grid_cell = np.mean([(g[1] - g[0]) for g in grs])
    dt        = grid_cell * 0.5
    t_span    = (0.0, dt * integration_steps)
    t_eval    = np.linspace(*t_span, integration_steps + 1)

    path_rows  = []
    arrow_rows = []

    for seed in seeds:
        try:
            sol = solve_ivp(
                _field, t_span, [seed[0], seed[1]],
                t_eval=t_eval, method="RK45", rtol=1e-3, atol=1e-4,
            )
        except Exception:
            continue

        xs = sol.y[0]
        ys = sol.y[1]

        dists_path, _ = cell_tree.query(np.column_stack([xs, ys]))
        valid_end = (
            int(np.argmax(dists_path > max_cell_dist))
            if np.any(dists_path > max_cell_dist)
            else len(xs)
        )
        if valid_end < 3:
            continue

        xs = xs[:valid_end]
        ys = ys[:valid_end]

        mid     = len(xs) // 2
        lw_here = float(interp_lw(np.array([[ys[mid], xs[mid]]])))
        lw_line = float(np.clip(
            linewidth * 2 * lw_here / (max_len + 1e-10),
            linewidth * 0.3,
            linewidth * 3.0,
        ))

        line_id = len(path_rows)
        for xi, yi in zip(xs, ys):
            path_rows.append({"x": float(xi), "y": float(yi), "line_id": line_id, "lw": lw_line})

        arrow_indices = list(range(arrow_every, len(xs) - 1, arrow_every))
        if not arrow_indices or arrow_indices[-1] < len(xs) - 2:
            arrow_indices.append(len(xs) - 2)

        for i in arrow_indices:
            arrow_rows.append({
                "x":    float(xs[i]),
                "y":    float(ys[i]),
                "xend": float(xs[i + 1]),
                "yend": float(ys[i + 1]),
            })

    if not path_rows:
        msg = (
            "stream(): no streamlines generated. "
            "Try raising max_cell_dist_frac, lowering min_mass, or raising density."
        )
        raise RuntimeError(msg)

    # BUILD: LetsPlot layers
    path_df  = pl.DataFrame(path_rows)
    arrow_df = pl.DataFrame(arrow_rows)

    path_data  = {c: path_df[c].to_list()  for c in path_df.columns}
    arrow_data = {c: arrow_df[c].to_list() for c in arrow_df.columns}

    layer_path = geom_path(
        data=path_data,
        mapping=aes(x="x", y="y", group="line_id", size="lw"),
        color=color,
        alpha=alpha,
    )

    layer_arrows = geom_segment(
        data=arrow_data,
        mapping=aes(x="x", y="y", xend="xend", yend="yend"),
        color=color,
        size=linewidth,
        alpha=alpha,
        arrow=arrow(type="closed", ends="last", length=arrow_size),
    )

    return layer_path + layer_arrows
dim + stream(dim)
```
