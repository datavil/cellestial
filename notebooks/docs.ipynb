```python
import scanpy as sc
import squidpy as sq
from lets_plot import *

import cellestial as cl

data = sc.read_h5ad("data/pbmc3k_pped.h5ad")
data_spatial = sc.read_h5ad("data/V1_Human_Lymph_Node_pped.h5ad")
data_hne = sq.datasets.visium_hne_adata()
```

```python
s4 = cl.spatial(data_hne, key="Mef2c") + scale_color_viridis(option="inferno")
s4
```

```python
s1 = cl.spatial(data_spatial, key="clusters")
s2 = cl.spatial(data_hne, key="leiden")
s3 = cl.spatial(data_spatial, key="MS4A1") + scale_color_viridis(option="inferno")
s4 = cl.spatial(data_hne, key="Mef2c") + scale_color_viridis(option="inferno")

gggrid([s1, s2, s3, s4],ncol=2) + ggsize(1000,800)
```

```python
dims = gggrid(
    [
        cl.umap(data, key="cell_type_lvl1", axis_type="arrow", size=1.5, legend_ondata=True)+scale_color_hue(),
        cl.tsne(
            data,
            key="NEAT1",
            tooltips=["NEAT1","cell_type_lvl1"],
            axis_type="arrow",
            size=2,
            color_high="#219B9D",
            color_low="#f6f6f6",
        ),
    ]
   ) + ggtb(size_zoomin=-1)
dims
```

```python
markers = ["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"]
htmp = cl.heatmap(
    data,
    group_by="cell_type_lvl1",
    keys=markers,
    geom="raster",
    group_lines_size=0.5,
    group_lines_color="white",
    dendrogram=True,
    dendrogram_size="0.7",
    group_bars_labels=True,
    group_bars=True,
) + scale_fill_viridis()

mtrx = cl.matrixplot(
    data,
    group_by="cell_type_lvl1",
    keys=markers,
    group_lines_size=0.5,
    group_lines_color="white",
    dendrogram=True,
    dendrogram_size="0.7",
)

mtrx = cl.matrixplot(
    data,
    group_by="cell_type_lvl1",
    keys=markers,
    group_lines_size=0.5,
    group_lines_color="white",
    dendrogram=True,
    dendrogram_size="0.7",
)

dtplt = cl.dotplot(
    data,
    group_by="leiden_res_0.50",
    keys=markers,
    group_lines_size=0.5,
    dendrogram=True,
    dendrogram_size="0.7",
)

stck = cl.stacked_violin(
    data,
    group_by="leiden_res_0.50",
    keys=markers,
    dendrogram=True,
    dendrogram_size="0.7",
)

gggrid(
    [
        htmp,
        mtrx,
        dtplt,
        stck,
    ],
    ncol=1,
    align=True
)  + ggsize(600,1400) + ggtb(size_zoomin=-1)
```
