```python
import scanpy as sc
from lets_plot import *

import cellestial as cl

data = sc.read_h5ad("data/pbmc3k_pped.h5ad")
```

```python
markers = ["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"]
```

```python
heat = cl.heatmap(
    data,
    keys = ["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"],
    group_by = "cell_type_lvl1",
    aggregate=False,
    rectangle=False,
    dendrogram=True,
    scale_axis=0,
    group_bars_labels = True,
    dendrogram_size=2
) 
heat+ scale_y_reverse() 
```

```python
heat = cl.heatmap(
    data,
    keys = ["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"],
    group_by = "cell_type_lvl1",
    rectangle=True,
    dendrogram=True,
    scale_axis=1
)
heat
```

```python
dot = cl.dotplot(
    data,
    keys=markers,
    group_by="cell_type_lvl1",
    dendrogram=True,
    dendrogram_color="gray",
    dendrogram_size=1,
    rectangle_color="gray",
    rectangle_size=3,
)
dot
```

```python
cl.retrieve(dot)
```

```python
sc.pl.dotplot(
    data,
    var_names=markers,
    groupby="cell_type_lvl1",
    dendrogram=True,
)
```
