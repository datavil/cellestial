```python
import scanpy as sc
import squidpy as sq
from lets_plot import *

import cellestial as cl

data = sc.read_h5ad("data/pbmc3k_pped.h5ad")
```

```python
markers = ["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"]
htmp = (
    cl.heatmap(
        data,
        group_by="cell_type_lvl1",
        keys=markers,
        geom="raster",
        group_lines_size=0.5,
        group_lines_color="white",
        dendrogram=True,
        dendrogram_size=0.7,
        group_bars_labels=True,
        group_bars=True,
        group_bars_size=3,
    )
    + scale_fill_viridis()
)
```

```python
cl.setup_html(isolated_frame=False)
```

```python
htmp + ggsize(800, 400) + cl.borders() + coord_fixed(ratio=0.5)
```

```python
(
    htmp
    + ggsize(2000, 1000)
    + cl.borders()
    + geom_point(data=dict(x=[0, -0.5, -1], y=[0, 0, 0]), mapping=aes(x="x", y="y"), size=5)
)
```
